```python
def make_chunk(
        chunk_id, 
        embedding, 
        text="t", 
        fund_name="Fund X",
        source_file="f.pdf", 
        page=1
    ) -> Chunk:

    return Chunk(
            chunk_id=chunk_id, 
            text=text, 
            embedding=embedding, 
            fund_name=fund_name, 
            source_file=source_file,
            page=page,
            metadata={}
        )
    

class _FakeEmbedder:

    def __init__(self, vector):
        self._vector = vector

    def embed_one(self, question) -> list[float]:
        return self._vector

class _FakeStore:

    def __init__(self, hits):
        self._hits = hits

    def search(self, query_embedding, top_k=5) -> list[Chunk]:
        return self._hits[:top_k]
```

```python
hits = [make_chunk("s1", [1, 0]), make_chunk("s2", [0, 1])]
fe = _FakeEmbedder([1,0])
fs = _FakeStore(hits)
```

```python

retr = Retriever(store=fs, embedder=fe)
out = retr.retrieve("any question", top_k=1)
out
```
